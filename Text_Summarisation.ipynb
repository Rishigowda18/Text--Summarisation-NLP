```python
# TEXT SUMMARISATION USING NLP
# Extractive Summarisation with TF-IDF

!pip install scikit-learn -q

import re
from sklearn.feature_extraction.text import TfidfVectorizer

# Input Text
text = """
Artificial Intelligence is one of the fastest growing technologies
in the modern world. It enables machines to perform tasks that
normally require human intelligence.

AI is widely used in healthcare, education, banking, transportation
and many other industries. In healthcare, AI helps doctors analyse
medical reports and detect diseases. In education, it supports
personalised learning and improves student engagement.

Natural Language Processing is an important branch of Artificial
Intelligence. It allows computers to understand, process and
generate human language. Text summarisation is one of the important
applications of NLP.

Text summarisation converts lengthy documents into shorter versions
while preserving their main meaning. It helps users save time and
understand important information quickly.

There are two main types of text summarisation: extractive and
abstractive summarisation. Extractive summarisation selects important
sentences from the original text, whereas abstractive summarisation
generates new sentences using the main ideas.

Text summarisation is useful for news articles, research papers,
educational materials, business reports and legal documents.
"""

# Step 1: Split text into sentences
sentences = re.split(r'(?<=[.!?])\s+', text.strip())
sentences = [s.strip() for s in sentences if s.strip()]

# Step 2: Convert sentences into TF-IDF vectors
vectorizer = TfidfVectorizer(stop_words='english')
tfidf_matrix = vectorizer.fit_transform(sentences)

# Step 3: Calculate sentence importance
scores = tfidf_matrix.sum(axis=1).A1

# Step 4: Select important sentences
summary_count = min(5, max(3, round(len(sentences) * 0.35)))

top_indices = scores.argsort()[-summary_count:]
top_indices = sorted(top_indices)

# Step 5: Generate summary
summary = " ".join(sentences[i] for i in top_indices)

# Step 6: Display results
print("=" * 55)
print("       TEXT SUMMARISATION USING NLP")
print("=" * 55)

print("\nORIGINAL TEXT:\n")
print(text)

print("\n" + "=" * 55)
print("SUMMARISED TEXT:\n")
print(summary)

print("\n" + "=" * 55)
print("SUMMARY STATISTICS")
print("Original sentences:", len(sentences))
print("Summary sentences:", summary_count)
print("Original words:", len(text.split()))
print("Summary words:", len(summary.split()))
print("=" * 55)
```
